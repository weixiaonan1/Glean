```python
import csv
import os

lines = []
data_dir = './data/banking/'
banking_files = ['train.tsv', 'dev.tsv', 'test.tsv']

for filename in banking_files:
    with open(os.path.join(data_dir, filename), "r") as f:
        reader = csv.reader(f, delimiter="\t")
        for i, line in enumerate(reader):
            if i == 0:
                continue
            line = [l.lower() for l in line][0]
            lines.append(line)

print(len(lines))
print(lines[0])
```
